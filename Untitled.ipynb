# Лабораторная работа №2

## Вариант: Искусство

В работе используется датасет произведений искусства. Числовой признак X — `width_mm`, ширина произведения в миллиметрах. Группирующий признак G — десятилетие, полученное из столбца `acquired`.

```python
SEED = 42
rng = np.random.default_rng(SEED)
```

```python
artworks = pd.read_csv("records.csv")
artworks.head()
```

Таблица загружена в DataFrame `artworks`. Метод `head()` показывает первые пять строк для первичной проверки данных. В таблице 8000 строк и 9 исходных столбцов.

```python
artworks.info()
```

```python
artworks["decade"] = artworks["acquired"] // 10 * 10
```

```python
artworks["decade"] = artworks["decade"].fillna("Неизвестно")
```

Десятилетие получено операцией `acquired // 10 * 10`: например, 1988 превращается в 1980. Пустые значения `acquired` собраны в группу `Неизвестно`, как требуется в задании.

```python
rows_with_width = artworks[artworks["width_mm"].notna()]
removed_rows = rows_with_width.sample(frac=0.1, random_state=SEED)
original_widths = removed_rows["width_mm"].copy()
artworks_with_missing = artworks.copy()
artworks_with_missing.loc[removed_rows.index, "width_mm"] = np.nan
```

```python
mean_width = artworks_with_missing["width_mm"].mean()
median_width = artworks_with_missing["width_mm"].median()
artworks_mean = artworks_with_missing.copy()
artworks_median = artworks_with_missing.copy()
artworks_interpolated = artworks_with_missing.copy()
```

```python
artworks_mean["width_mm"] = artworks_mean["width_mm"].fillna(mean_width)
artworks_median["width_mm"] = artworks_median["width_mm"].fillna(median_width)
artworks_interpolated["width_mm"] = artworks_interpolated["width_mm"].interpolate()
artworks_interpolated["width_mm"] = artworks_interpolated["width_mm"].fillna(median_width)
```

```python
mean_restored_widths = artworks_mean.loc[removed_rows.index, "width_mm"]
mean_errors = mean_restored_widths - original_widths
mean_absolute_error = np.abs(mean_errors).mean()
mean_squared_error = (mean_errors ** 2).mean()
print("MAE:", mean_absolute_error)
print("MSE:", mean_squared_error)
```

```python
median_restored_widths = artworks_median.loc[removed_rows.index, "width_mm"]
median_errors = median_restored_widths - original_widths
median_absolute_error = np.abs(median_errors).mean()
median_squared_error = (median_errors ** 2).mean()
print("MAE:", median_absolute_error)
print("MSE:", median_squared_error)
```

```python
interpolated_restored_widths = artworks_interpolated.loc[removed_rows.index, "width_mm"]
interpolated_errors = interpolated_restored_widths - original_widths
interpolated_absolute_error = np.abs(interpolated_errors).mean()
interpolated_squared_error = (interpolated_errors ** 2).mean()
print("MAE:", interpolated_absolute_error)
print("MSE:", interpolated_squared_error)
```

Для проверки были удалены 10% заполненных значений `width_mm`. По MAE лучше медиана, по MSE лучше среднее. Интерполяция в этом датасете дала худший результат. MSE сильнее реагирует на большие ошибки, потому что возводит их в квадрат.

```python
artworks_for_duplicates = artworks_median.copy()
rows_before_duplicates = len(artworks_for_duplicates)
average_before_duplicates = artworks_for_duplicates["width_mm"].mean()
print("Строк до добавления:", rows_before_duplicates)
print("Средняя ширина до добавления:", average_before_duplicates)
```

```python
duplicate_rows = artworks_for_duplicates.sample(frac=0.05, random_state=SEED)
artworks_with_duplicates = pd.concat(
    [artworks_for_duplicates, duplicate_rows],
    ignore_index=True)
print("Строк после добавления:", len(artworks_with_duplicates))
print("Средняя ширина после добавления:", artworks_with_duplicates["width_mm"].mean())
```

```python
duplicates_by_all_columns = artworks_with_duplicates.copy()
duplicates_by_id = artworks_with_duplicates.copy()
duplicates_by_url = artworks_with_duplicates.copy()
```

```python
without_duplicates_by_all_columns = duplicates_by_all_columns.drop_duplicates()
print("Строк после удаления:", len(without_duplicates_by_all_columns))
print("Средняя ширина:", without_duplicates_by_all_columns["width_mm"].mean())
```

```python
without_duplicates_by_id = duplicates_by_id.drop_duplicates(subset="id")
print("Строк после удаления по id:", len(without_duplicates_by_id))
print("Средняя ширина:", without_duplicates_by_id["width_mm"].mean())
```

```python
without_duplicates_by_url = duplicates_by_url.drop_duplicates(subset="url")
print("Строк после удаления по url:", len(without_duplicates_by_url))
print("Средняя ширина:", without_duplicates_by_url["width_mm"].mean())
```

```python
url_id_counts = duplicates_by_url.groupby("url")["id"].nunique()
different_ids_for_same_url = url_id_counts[url_id_counts > 1]
print("Количество ссылок с разными id:", len(different_ids_for_same_url))
```

После добавления 5% точных копий стало 8400 строк. После удаления повторов каждым способом осталось 8000 строк. Одинаковых URL с разными ID не найдено.

```python
width_values = artworks["width_mm"].dropna().tolist()
print("Количество значений:", len(width_values))

smallest_width = width_values[0]
largest_width = width_values[0]
total_width = 0

for width in width_values:
    if width < smallest_width:
        smallest_width = width
    if width > largest_width:
        largest_width = width
    total_width = total_width + width

width_range = largest_width - smallest_width
average_width = total_width / len(width_values)

print("Минимум:", smallest_width)
print("Максимум:", largest_width)
print("Размах:", width_range)
print("Сумма:", total_width)
print("Количество:", len(width_values))
print("Среднее:", average_width)
```

```python
sorted_widths = sorted(width_values)
middle_index = len(sorted_widths) // 2

if len(sorted_widths) % 2 == 1:
    median_width = sorted_widths[middle_index]
else:
    median_width = (sorted_widths[middle_index - 1] + sorted_widths[middle_index]) / 2

print("Медиана:", median_width)
```

```python
sum_squared_deviations = 0

for width in width_values:
    difference = width - average_width
    sum_squared_deviations = sum_squared_deviations + difference ** 2

manual_variance = sum_squared_deviations / len(width_values)
manual_standard_deviation = manual_variance ** 0.5

print("Дисперсия:", manual_variance)
print("Стандартное отклонение:", manual_standard_deviation)
```

```python
width_array = np.array(width_values)
width_series = artworks["width_mm"].dropna()

print("Наши расчёты:")
print("Размах:", width_range)
print("Среднее:", average_width)
print("Медиана:", median_width)
print("Дисперсия:", manual_variance)
print("Стандартное отклонение:", manual_standard_deviation)

print("\nNumPy:")
print("Размах:", np.ptp(width_array))
print("Среднее:", np.mean(width_array))
print("Медиана:", np.median(width_array))
print("Дисперсия:", np.var(width_array))
print("Стандартное отклонение:", np.std(width_array))

print("\npandas:")
print("Среднее:", width_series.mean())
print("Медиана:", width_series.median())
print("Дисперсия:", width_series.var(ddof=0))
print("Стандартное отклонение:", width_series.std(ddof=0))
```

Ручные расчёты совпали с NumPy и pandas. Размах равен 8907 мм, среднее — 323.71 мм, медиана — 175 мм, дисперсия — 177040.12, стандартное отклонение — 420.76 мм. Среднее больше медианы из-за очень больших значений ширины.

```python
chosen_decades = [1850, 1900]
width_limit = 300

is_chosen_decade = artworks["decade"].isin(chosen_decades)
is_wide = artworks["width_mm"] > width_limit

wide_in_chosen_decades = artworks[is_chosen_decade & is_wide]
chosen_decades_or_wide = artworks[is_chosen_decade | is_wide]

print("Первая таблица:", len(wide_in_chosen_decades))
print("Вторая таблица:", len(chosen_decades_or_wide))
```

```python
only_in_second = chosen_decades_or_wide[
    ~chosen_decades_or_wide.index.isin(wide_in_chosen_decades.index)
]

print("Только во второй таблице:", len(only_in_second))
only_in_second.head()
```

Для фильтрации выбраны десятилетия 1850 и 1900 и граница 300 мм. Условие «и» дало 213 строк, условие «или» — 6331 строку. Отсутствующее значение ширины не считается значением больше 300.

```python
suitable_rows = is_chosen_decade & is_wide
filtered_first = artworks[suitable_rows]
top_five_after_filter = filtered_first.sort_values(
    "width_mm",
    ascending=False
).head(5)

top_five_first = artworks.sort_values(
    "width_mm",
    ascending=False
).head(5)

top_five_then_filter = top_five_first[
    top_five_first["decade"].isin(chosen_decades)
    & (top_five_first["width_mm"] > width_limit)
]

print("Сначала фильтр, потом пять строк:")
print(top_five_after_filter[["id", "decade", "width_mm"]])

print("Сначала пять строк, потом фильтр:")
print(top_five_then_filter[["id", "decade", "width_mm"]])
```

Порядок операций влияет на результат: после предварительной фильтрации нашлись пять строк, а после выбора пяти самых широких строк всей таблицы подходящих строк не осталось. Результаты не всегда будут одинаковыми.

```python
group_summary = artworks.groupby("decade")["width_mm"].agg(
    ["size", "count", "mean", "median"])
group_summary["missing"] = (
    group_summary["size"] - group_summary["count"])
group_summary = group_summary.rename(
    columns={
        "size": "objects",
        "mean": "average_width",
        "median": "median_width"})
group_summary["share"] = group_summary["objects"] / len(artworks)
group_summary
```

Для каждой группы посчитаны количество строк, доля, число пропусков, среднее и медиана. Самая большая группа — 1850 год: 4397 произведений. Пустые даты находятся в группе `Неизвестно`.

```python
groups_for_chart = group_summary.sort_values("objects")
plt.figure(figsize=(9, 6))
plt.barh(
    groups_for_chart.index.astype(str),
    groups_for_chart["objects"])
plt.xlabel("Количество произведений")
plt.ylabel("Десятилетие приобретения")
plt.title("Количество произведений по десятилетиям")
plt.show()
```

На горизонтальной диаграмме самая большая группа находится сверху, а самая маленькая — снизу. Больше всего произведений относится к 1850 году.

```python
group_counts = artworks["decade"].value_counts()
group_means = artworks.groupby("decade")["width_mm"].mean()
artworks["group_count"] = artworks["decade"].map(group_counts)
artworks["group_mean"] = artworks["decade"].map(group_means)
artworks["delta"] = artworks["width_mm"] - artworks["group_mean"]
print("Количество строк:", len(artworks))

artworks[
    ["id", "decade", "width_mm", "group_count", "group_mean", "delta"]
].head()
```

`group_count` показывает размер группы, `group_mean` — среднюю ширину группы, а `delta` равен `width_mm - group_mean`. Положительное `delta` означает, что произведение шире среднего своей группы.

```python
first_100 = artworks[artworks["width_mm"].notna()].head(100).copy()
first_100["object_number"] = range(1, len(first_100) + 1)
plt.figure(figsize=(10, 5))
plt.scatter(first_100["object_number"], first_100["delta"])
plt.axhline(0, color="red", linestyle="--")
plt.xlabel("Порядковый номер произведения")
plt.ylabel("Отклонение от среднего группы")
plt.title("Отклонения первых 100 произведений")
plt.show()
```

На точечном графике значения выше нулевой линии больше среднего своей группы, а значения ниже линии меньше среднего.

```python
sorted_by_delta = first_100.sort_values("delta").copy()
sorted_by_delta["new_number"] = range(
    1, len(sorted_by_delta) + 1)
plt.figure(figsize=(10, 5))
plt.plot(
    sorted_by_delta["new_number"],
    sorted_by_delta["delta"],
    marker="o")
plt.xlabel("Новый порядковый номер")
plt.ylabel("Отклонение от среднего группы")
plt.title("Отсортированные отклонения")
plt.show()
```

После сортировки по возрастанию `delta` линия идёт слева направо без убывания, как требует задание.

```python
width_quantiles = artworks["width_mm"].dropna().quantile(
    [0.25, 0.5, 0.75])
print(width_quantiles)
```

Квантили равны: 0.25 — 119 мм, 0.5 — 175 мм, 0.75 — 337 мм. Центральная половина значений находится между 119 и 337 мм.

```python
largest_groups = group_summary.sort_values(
    "objects",
    ascending=False).head(5).index
boxplot_values = []

for group in largest_groups:
    group_widths = artworks.loc[
        artworks["decade"] == group,
        "width_mm"].dropna()

    boxplot_values.append(group_widths)

plt.figure(figsize=(9, 6))

plt.boxplot(
    boxplot_values,
    tick_labels=[str(group) for group in largest_groups])

plt.xlabel("Десятилетие приобретения")
plt.ylabel("Ширина произведения, мм")
plt.title("Распределение ширины в пяти самых больших группах")
plt.show()
```

В boxplot линия внутри ящика показывает медиану, ящик — центральную половину значений, а отдельные точки — выбросы. Самая высокая медиана у группы 2000 года; сильнее всего значения разбросаны в группах 2000 и 1970 годов.
