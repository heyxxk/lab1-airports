# Лабораторная работа №2

## Вариант: Искусство

В работе используется датасет произведений искусства. Числовой признак X — `width_mm`, ширина произведения в миллиметрах. Группирующий признак G — десятилетие, полученное из столбца `acquired`.

## 1. Подготовка

Подключаем библиотеки и задаём `SEED`, чтобы случайный выбор можно было повторить.

```python
SEED = 42
rng = np.random.default_rng(SEED)
```

## 2. Загрузка данных

Загружаем CSV в таблицу pandas и проверяем первые строки, столбцы, типы данных и пропуски.

```python
artworks = pd.read_csv("records.csv")
artworks.head()
```

Таблица загружена в DataFrame `artworks`. Метод `head()` показывает первые пять строк для первичной проверки данных. В таблице 8000 строк и 9 исходных столбцов.

```python
artworks.info()
```

```python
artworks["decade"] = artworks["acquired"] // 10 * 10
```

```python
artworks["decade"] = artworks["decade"].fillna("Неизвестно")
```

Десятилетие получено операцией `acquired // 10 * 10`: например, 1988 превращается в 1980. Пустые значения `acquired` собраны в группу `Неизвестно`, как требуется в задании.

## 3. Пропуски

Создаём одинаковые искусственные пропуски и восстанавливаем их средним, медианой и интерполяцией.

```python
available_widths = artworks[artworks["width_mm"].notna()]
removed = available_widths.sample(frac=0.1, random_state=SEED)
removed_index = removed.index
original_widths = removed["width_mm"].copy()

missing_data = artworks.copy()
missing_data.loc[removed_index, "width_mm"] = np.nan
```

```python
mean_value = missing_data["width_mm"].mean()
median_value = missing_data["width_mm"].median()

mean_table = missing_data.copy()
median_table = missing_data.copy()
interpolation_table = missing_data.copy()
```

```python
mean_table["width_mm"] = mean_table["width_mm"].fillna(mean_value)
median_table["width_mm"] = median_table["width_mm"].fillna(median_value)
interpolation_table["width_mm"] = interpolation_table["width_mm"].interpolate().fillna(median_value)

errors_mean = mean_table.loc[removed_index, "width_mm"] - original_widths
errors_median = median_table.loc[removed_index, "width_mm"] - original_widths
errors_interpolation = interpolation_table.loc[removed_index, "width_mm"] - original_widths

print("Среднее: MAE =", np.abs(errors_mean).mean(), "MSE =", (errors_mean ** 2).mean())
print("Медиана: MAE =", np.abs(errors_median).mean(), "MSE =", (errors_median ** 2).mean())
print("Интерполяция: MAE =", np.abs(errors_interpolation).mean(), "MSE =", (errors_interpolation ** 2).mean())
```

Для проверки были удалены 10% заполненных значений `width_mm`. По MAE лучше медиана, по MSE лучше среднее. Интерполяция в этом датасете дала худший результат. MSE сильнее реагирует на большие ошибки, потому что возводит их в квадрат.

## 4. Дубликаты

Добавляем 5% точных копий, затем проверяем три способа удаления повторов.

```python
duplicate_source = median_table.copy()
rows_before = len(duplicate_source)
mean_before = duplicate_source["width_mm"].mean()

print("До добавления копий:", rows_before, mean_before)
```

```python
copies = duplicate_source.sample(frac=0.05, random_state=SEED)
with_duplicates = pd.concat([duplicate_source, copies], ignore_index=True)

print("После добавления копий:", len(with_duplicates), with_duplicates["width_mm"].mean())
```

```python
without_duplicates_all = with_duplicates.drop_duplicates()
without_duplicates_id = with_duplicates.drop_duplicates(subset="id")
without_duplicates_url = with_duplicates.drop_duplicates(subset="url")

print("По всем столбцам:", len(without_duplicates_all), without_duplicates_all["width_mm"].mean())
print("По id:", len(without_duplicates_id), without_duplicates_id["width_mm"].mean())
print("По url:", len(without_duplicates_url), without_duplicates_url["width_mm"].mean())

url_id_counts = with_duplicates.groupby("url")["id"].nunique()
urls_with_different_ids = url_id_counts[url_id_counts > 1]
print("Ссылок с разными id:", len(urls_with_different_ids))
```

После добавления 5% точных копий стало 8400 строк. После удаления повторов каждым способом осталось 8000 строк. Одинаковых URL с разными ID не найдено.

## 5. Числовые характеристики

Вручную считаем основные характеристики ширины и сравниваем их с NumPy и pandas.

```python
widths = artworks["width_mm"].dropna().tolist()
widths.sort()
number_of_widths = len(widths)

smallest = widths[0]
largest = widths[-1]
width_range = largest - smallest

total = 0
for width in widths:
    total += width
mean_width = total / number_of_widths

if number_of_widths % 2 == 1:
    median_width = widths[number_of_widths // 2]
else:
    middle = number_of_widths // 2
    median_width = (widths[middle - 1] + widths[middle]) / 2

squared_deviations = 0
for width in widths:
    squared_deviations += (width - mean_width) ** 2

variance = squared_deviations / number_of_widths
standard_deviation = variance ** 0.5

print("Размах:", width_range)
print("Среднее:", mean_width)
print("Медиана:", median_width)
print("Дисперсия:", variance)
print("Стандартное отклонение:", standard_deviation)
```

Ручные расчёты совпали с NumPy и pandas. Размах равен 8907 мм, среднее — 323.71 мм, медиана — 175 мм, дисперсия — 177040.12, стандартное отклонение — 420.76 мм. Среднее больше медианы из-за очень больших значений ширины.

## 6. Фильтрация

Выбираем два десятилетия и сравниваем ширину с границей 300 мм.

```python
chosen_decades = [1850, 1900]
width_limit = 300

is_chosen_decade = artworks["decade"].isin(chosen_decades)
is_wide = artworks["width_mm"] > width_limit

wide_in_chosen_decades = artworks[is_chosen_decade & is_wide]
chosen_decades_or_wide = artworks[is_chosen_decade | is_wide]

print("Условие И:", len(wide_in_chosen_decades))
print("Условие ИЛИ:", len(chosen_decades_or_wide))
```

```python
only_in_second = chosen_decades_or_wide[
    ~chosen_decades_or_wide.index.isin(wide_in_chosen_decades.index)
]

print("Только во второй таблице:", len(only_in_second))
only_in_second.head()
```

Для фильтрации выбраны десятилетия 1850 и 1900 и граница 300 мм. Условие «и» дало 213 строк, условие «или» — 6331 строку. Отсутствующее значение ширины не считается значением больше 300.

## 7. Порядок операций

Сравниваем результат, если сначала фильтровать строки, и результат, если сначала брать пять самых больших значений.

```python
filtered_rows = artworks[is_chosen_decade & is_wide]
top_after_filter = filtered_rows.sort_values("width_mm", ascending=False).head(5)

top_all = artworks.sort_values("width_mm", ascending=False).head(5)
top_then_filter = top_all[top_all["decade"].isin(chosen_decades) & (top_all["width_mm"] > width_limit)]

print("Сначала фильтр, потом топ-5:")
print(top_after_filter[["id", "decade", "width_mm"]])
print("Сначала топ-5, потом фильтр:")
print(top_then_filter[["id", "decade", "width_mm"]])
```

Порядок операций влияет на результат: после предварительной фильтрации нашлись пять строк, а после выбора пяти самых широких строк всей таблицы подходящих строк не осталось. Результаты не всегда будут одинаковыми.

## 8. Группировка

Для каждого десятилетия считаем количество, долю, пропуски, среднее и медиану и строим диаграмму.

```python
groups = artworks.groupby("decade")

group_summary = pd.DataFrame({
    "objects": groups.size(),
    "share": groups.size() / len(artworks),
    "missing": groups.size() - groups["width_mm"].count(),
    "average_width": groups["width_mm"].mean(),
    "median_width": groups["width_mm"].median()
})

group_summary
```

Для каждой группы посчитаны количество строк, доля, число пропусков, среднее и медиана. Самая большая группа — 1850 год: 4397 произведений. Пустые даты находятся в группе `Неизвестно`.

```python
groups_for_chart = group_summary.sort_values("objects")
plt.figure(figsize=(9, 6))
plt.barh(
    groups_for_chart.index.astype(str),
    groups_for_chart["objects"])
plt.xlabel("Количество произведений")
plt.ylabel("Десятилетие приобретения")
plt.title("Количество произведений по десятилетиям")
plt.show()
```

На горизонтальной диаграмме самая большая группа находится сверху, а самая маленькая — снизу. Больше всего произведений относится к 1850 году.

## 9. Объект и его группа

Добавляем к каждой строке размер группы, среднее группы и отклонение `delta`.

```python
group_counts = groups.size()
group_means = groups["width_mm"].mean()

artworks["group_count"] = artworks["decade"].map(group_counts)
artworks["group_mean"] = artworks["decade"].map(group_means)
artworks["delta"] = artworks["width_mm"] - artworks["group_mean"]

print("Количество строк:", len(artworks))
artworks[["id", "decade", "width_mm", "group_count", "group_mean", "delta"]].head()
```

`group_count` показывает размер группы, `group_mean` — среднюю ширину группы, а `delta` равен `width_mm - group_mean`. Положительное `delta` означает, что произведение шире среднего своей группы.

```python
first_100 = artworks[artworks["width_mm"].notna()].head(100)

plt.figure(figsize=(10, 5))
plt.scatter(range(1, len(first_100) + 1), first_100["delta"])
plt.axhline(0, color="red", linestyle="--")
plt.xlabel("Номер произведения")
plt.ylabel("Отклонение от среднего группы")
plt.title("Отклонения первых 100 произведений")
plt.show()
```

На точечном графике значения выше нулевой линии больше среднего своей группы, а значения ниже линии меньше среднего.

```python
sorted_by_delta = first_100.sort_values("delta")

plt.figure(figsize=(10, 5))
plt.plot(range(1, len(sorted_by_delta) + 1), sorted_by_delta["delta"], marker="o")
plt.xlabel("Новый порядковый номер")
plt.ylabel("Отклонение от среднего группы")
plt.title("Отсортированные отклонения")
plt.show()
```

После сортировки по возрастанию `delta` линия идёт слева направо без убывания, как требует задание.

## 10. Квантили и boxplot

Считаем квантили и сравниваем распределение ширины в пяти самых больших группах.

```python
width_quantiles = artworks["width_mm"].dropna().quantile([0.25, 0.5, 0.75])
print(width_quantiles)
```

Квантили равны: 0.25 — 119 мм, 0.5 — 175 мм, 0.75 — 337 мм. Центральная половина значений находится между 119 и 337 мм.

```python
largest_groups = group_summary.sort_values("objects", ascending=False).head(5).index
boxplot_values = []

for group in largest_groups:
    values_in_group = artworks.loc[artworks["decade"] == group, "width_mm"].dropna()
    boxplot_values.append(values_in_group)

plt.figure(figsize=(9, 6))
plt.boxplot(boxplot_values, tick_labels=[str(group) for group in largest_groups])
plt.xlabel("Десятилетие приобретения")
plt.ylabel("Ширина произведения, мм")
plt.title("Распределение ширины в пяти самых больших группах")
plt.show()
```

В boxplot линия внутри ящика показывает медиану, ящик — центральную половину значений, а отдельные точки — выбросы. Самая высокая медиана у группы 2000 года; сильнее всего значения разбросаны в группах 2000 и 1970 годов.
